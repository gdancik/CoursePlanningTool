## Example code for reading a sheet using the /api/getSheet endpoing

Here we read a sheet using the PythonAnywhere API.

```python
import requests
import json
import pandas as pd

baseurl = 'http://127.0.0.1:5000/'
#baseurl = 'https://gdancik.pythonanywhere.com/'

session = requests.Session()
response = session.get(baseurl + 'api/test_login/?user=annie&password=password')
```

```python

response = session.post(baseurl + 'api/getSheet/')
sheet = json.loads(response.text)
sheet = pd.DataFrame.from_dict(sheet, orient = 'index')
sheet
```

```python
course_id = 'XHNBYqqdPjLgLWxteLbB'

response = session.post(baseurl + 'api/getCourse/',
                        json = {'course_id': course_id}
)
json.loads(response.text)
```

```python
response = session.post(baseurl + 'api/getValue/',
                        json = {'course_id': 'XHNBYqqdPjLgLWxteLbB', 
                                'list_of_columns' : ['instructor_name_syllabus', 'additional_sections_syllabus_json']}
)
json.loads(response.text)
```

```python
response = session.post(baseurl + 'api/deleteCourse/',
                        json = {'course_id': 'zdrKSHMmwigvDGtOwUSn'}
                                
)
json.loads(response.text)
```
