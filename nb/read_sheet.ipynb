## Example code for reading a sheet using the /api/getSheet endpoing

Here we read a sheet using the PythonAnywhere API.

```python
import requests
import json
import pandas as pd

session = requests.Session()
response = session.get('https://gdancik.pythonanywhere.com/api/test_login/?user=annie&password=password')
response = session.post('https://gdancik.pythonanywhere.com/api/getSheet/')
sheet = json.loads(response.text)
sheet = pd.DataFrame.from_dict(sheet)
sheet
```
